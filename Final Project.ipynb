## 3.Data Manipulation

### 3.1 Merge Data

```python
df = customers.merge(
    loans,
    on = "customer_id",
    how = "inner"
)

df = df.merge (
    loan_payments,
    on = "loan_id",
    how = "left"
)
```

### 3.2 Create a New Colomns

```python
#membuat kolom sisa pinjaman
df['remaining_balance'] = (df['loan_amount']- df.groupby('loan_id')['principal_component'].transform('sum'))
df.loc[df['remaining_balance'] <=0, 'remaining_balance'] = 0     # Jika sisa pinjaman 0 maka status close/ cicilan sudah lunas

#membuat kolom umur
df['age'] = (pd.to_datetime('today') - df['date_of_birth']).dt.days // 365

#membuat kolom credit kategori
bins = [299,579,669,739,799,899]
labels =['very poor','poor','fair','good','excellent']
df['credit_category']=pd.cut(df['credit_score'], bins = bins, labels = labels)

#membuat kolom jatuh tempo 30 hari setelah start date
df['due_date'] = df['start_date'] + pd.DateOffset(months = 1)

df.head()
```

##5 Clustering Risk Customer

###5.1 Clutering Income

```python
#clustering income
customer_income = df.groupby("customer_id").agg(
    avg_income = ("annual_income","mean"),
    total_income = ("annual_income","sum"),
    income_count = ("annual_income","count")
).reset_index()
```

```python
income_features = [
    "avg_income",
    "total_income",
    "income_count"
]
x=customer_income[income_features].copy()
```

```python
#scaling
from sklearn.preprocessing import StandardScaler
scaler_loan = StandardScaler()
scaler_payment = StandardScaler()
scaler_income = StandardScaler()


x_loan_scaler= scaler_loan.fit_transform(x)
x_payment_scaler = scaler_payment.fit_transform(x)
x_income_scaler = scaler_income.fit_transform(x)
```

```python
#income model K- means
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

for k in range(2,11):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    model.fit(x_income_scaler)
    inertia.append(model.inertia_)
plt.plot(range(2,11), inertia, marker= "o")
plt.xlabel("Jumlah Cluster")
plt.ylabel("Inertia")
plt.title("Elbow_Method")
plt.show()
```

```python
#kmeans income
from sklearn.cluster import KMeans
kmeans = KMeans(
    n_clusters=3,
    n_init=10,
    random_state=42
)

customer_income["income_cluster"] = kmeans.fit_predict(x_income_scaler)
```

```python
#cluster income name
cluster_name={
    0 : "Low_Income",
    1 : "Medium_Income",
    2 : "High_Income"
}

customer_income["customer_segment"] = customer_income["income_cluster"].map(cluster_name)
```

```python
customer_income.head(10)
```

```python
plt.figure(figsize=(12,6))

ax= plt.subplot()
sns.countplot(data=customer_income, x="customer_segment", ax=ax)

ax.set_title("Clustering Income")
for container in ax.containers:
        ax.bar_label(
            container,
            fmt="%.0f",
            padding = 3
        )
ax.set_xlabel("Income Cluster")
ax.set_ylabel("Customer")
plt.tight_layout()
plt.show()
```

```python
#one hot encoding
df_encoded = pd.get_dummies(
    df[["customer_id","status","loan_type"]],
    columns=["status","loan_type"]
)
df_encoded.head()
```

```python
#Aggregasi level customer
customer_status_type = df_encoded.groupby("customer_id").sum().reset_index()
customer_status_type.head()
```

```python
#features
features = customer_status_type.drop(columns="customer_id")
```

```python
#scalling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_scaled =scaler.fit_transform(features)
```

```python
#elbow method

from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

for k in range(2, 10):
  kmeans =KMeans(
      n_clusters=k,
      random_state=42,
      n_init=10
  )
  kmeans.fit(x_scaled)
  inertia.append(kmeans.inertia_)

plt.plot(range(2,10), inertia, marker="o")
plt.xlabel("Jumlah Cluster")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

```python

conditions = [
    # 1. High Value Healthy
    (
        (all_cluster['customer_segment'] == 'High_Income') &
        (all_cluster['segment'].str.contains('High_Exposure_Borrows/High_Activity', na=False)) &
        (all_cluster['segment_payment'] == 'High_Payment_Activity') &
        (all_cluster['segment_status'] == 'Status_active')
    ),

    # 2. High Value Risky
    (
        (all_cluster['customer_segment'] == 'High_Income') &
        (all_cluster['segment'].str.contains('High_Exposure_Borrows/High_Activity', na=False)) &
        (all_cluster['segment_payment'] == 'High_Payment_Activity') &
        (all_cluster['segment_status'] == 'Status_defaulted') | (all_cluster['segment_status'] == "Status_Written_off")
    ),

    # 3. Medium Risky
    (
        (all_cluster['customer_segment'] == 'Medium_Income') &
        (all_cluster['segment_status'] == 'Status_defaulted') | (all_cluster['segment_status'] == "Status_Written_off")
    ),

    # 4. Low Income High Commitment
    (
        (all_cluster['customer_segment'] == 'Low_Income') &
        (all_cluster['segment'].str.contains('High_Exposure_Borrows/High_Activity', na=False)) &
        (all_cluster['segment_payment'] == 'High_Payment_Activity') &
        (all_cluster['segment_status'] == 'Status_active')
    ),

    # 5. Low Value Low Activity
    (
        (all_cluster['customer_segment'] == 'Low_Income') &
        (all_cluster['segment'].str.contains('Low_Exposure', na=False)) &
        (all_cluster['segment_payment'] == 'Low_Payment_Activity') &
        (all_cluster['segment_status'] == 'Status_active')
    ),

    # 6. Medium Stable
    (
        (all_cluster['customer_segment'] == 'Medium_Income') &
        (all_cluster['segment_status'] == 'Status_active')
    )
]

choices = [
    'High Value Healthy',
    'High Value Risky',
    'Medium Risky',
    'Low Income High Commitment',
    'Low Value Low Activity',
    'Medium Stable'
]

all_cluster['final_segment'] = np.select(
    conditions,
    choices,
    default='Other'
)
```
